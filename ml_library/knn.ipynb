```python
import pandas as pd
import numpy as np

def load_data(path):
    df = pd.read_csv(path)
    X = df.iloc[:, :-1].values.astype(float)
    y = df.iloc[:, -1].values
    return X, y

def normalize_features(X):
    mu = np.mean(X, axis=0)
    sigma = np.std(X, axis=0)
    sigma[sigma == 0] = 1
    X_norm = (X - mu) / sigma
    return X_norm, mu, sigma

class KNN:
    def __init__(self, k=5):
        self.k = k

    def fit(self, X, y):
        self.X_train = X
        self.y_train = y

    def predict(self, X):
        preds = []
        for x in X:
            distances = np.linalg.norm(self.X_train - x, axis=1)
            k_idx = np.argsort(distances)[1:self.k+1]  # skip self-distance
            labels = self.y_train[k_idx]

            values, counts = np.unique(labels, return_counts=True)
            preds.append(values[np.argmax(counts)])

        return np.array(preds)

def save_submission(predictions, filename="submission.csv", column_name="target"):
    predictions = predictions.reshape(-1)
    df = pd.DataFrame({column_name: predictions})
    df.to_csv(filename, index=False)
    print(f"Submission file saved as '{filename}'")

#To run the model
X, y = load_data("unsupervised_data.csv")
X, mu, sigma = normalize_features(X)

knn = KNN(k=5)
knn.fit(X, y)

y_pred = knn.predict(X)
print(y_pred[:5])

save_submission(y_pred, filename="knn_submission.csv")
```
