```python
import numpy as np
import pandas as pd
 
def create_submission(labels, filename="k_means_clust_pred.csv"):
    submission = pd.DataFrame({"id": np.arange(len(labels)), "cluster": labels})
    submission.to_csv(filename, index=False)
    print(f"Submission file saved as {filename}")

def load_data(path):
    df = pd.read_csv(path)
    return df.values

def normalize(X):
    mu = np.mean(X, axis=0)
    sigma = np.std(X, axis=0)
    return (X - mu) / sigma

def initialize_centroids(X, k, seed=42):
    np.random.seed(seed)
    indices = np.random.choice(X.shape[0], k, replace=False)
    return X[indices]

def assign_clusters(X, centroids):
    distances = np.zeros((X.shape[0], centroids.shape[0]))

    for i in range(centroids.shape[0]):
        distances[:, i] = np.linalg.norm(X - centroids[i], axis=1)
    return np.argmin(distances, axis=1)

def update_centroids(X, labels, k):
    new_centroids = np.zeros((k, X.shape[1]))

    for i in range(k):
        cluster_points = X[labels == i]
        if len(cluster_points) > 0:
            new_centroids[i] = np.mean(cluster_points, axis=0)

    return new_centroids

def k_means(X, k, max_iters=100, tol=1e-4):
    centroids = initialize_centroids(X, k)

    for iteration in range(max_iters):
        labels = assign_clusters(X, centroids)
        new_centroids = update_centroids(X, labels, k)

        shift = np.linalg.norm(new_centroids - centroids)

        if shift < tol:
            print(f"Converged at iteration {iteration}")
            break

        centroids = new_centroids

    return centroids, labels

#To run the model
X = load_data("unsupervised_data.csv")
X = normalize(X)
k = 3  
centroids, labels = k_means(X, k)
print("Final Centroids:\n", centroids)
print("Cluster Labels:\n", labels)
create_submission(labels)
```
